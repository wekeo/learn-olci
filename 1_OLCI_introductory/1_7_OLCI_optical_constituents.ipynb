## Reading OLCI optical consitutent data
[Back to top](#TOC_TOP_1_7)

</div>

To begin with, we'll create paths that point to the location of our data of interest.

```python
baseline_collection = '003'
SAFE_directory = glob.glob(os.path.join(os.getcwd(), 'products', f'*WFR*{baseline_collection}*'))[0]
```

Next we'll set a parameter to reduce the resolution of the data we are working with. For analyses you would not want to do this, but it will help us to visualise our data much faster in the context of this tutorial notebook.

```python
grid_reduce = 5
```

Then, we'll open our latitude and longitude grid, subsampling by `grid_reduce`.

```python
geo_file = os.path.join(SAFE_directory,'geo_coordinates.nc')
geo_fid = xr.open_dataset(geo_file)
lon = geo_fid.longitude.data[::grid_reduce, ::grid_reduce]
lat = geo_fid.latitude.data[::grid_reduce, ::grid_reduce]
geo_fid.close()
```

Lastly, we will open our three optical consistuent products, as provided by the complex water processor:
* the neural network chlorophyll product (CHL_NN; as discuss in the previous notebook)
* the neural network total suspended sediment product (TSM_NN)
* the neural network coloured detrital matter (CDM) absorption coefficient at 443 nm (ADG443_NN)

```python
band_var = {}
band_var["chl_nn"] = xr.open_mfdataset(glob.glob(os.path.join(SAFE_directory,'chl_nn.nc')))["CHL_NN"]
band_var["tsm_nn"] = xr.open_mfdataset(glob.glob(os.path.join(SAFE_directory,'tsm_nn.nc')))["TSM_NN"]
band_var["adg_443_nn"] = xr.open_mfdataset(glob.glob(os.path.join(SAFE_directory,'iop_nn.nc')))["ADG443_NN"]
```

<div class="alert alert-info" role="alert">

## Applying flags
[Back to top](#TOC_TOP_1_7)

</div>

As well as the data and the geocoordinates, to plot geophysical products derived from OLCI properly, we also need to extract and apply the relevant quality flags. Below you can see that the flags are extracted from the wqsf netcdf file. If we print out the flags and their associated bit values, you can get an idea about what they represent, and how to access each individual one.

```python
flag_file = os.path.join(SAFE_directory,'wqsf.nc')
```

We'll use a quick function from the eumar tools library that allows us to read in the flag information simply by supplying the name. For our recommendations on which flags to apply you should also consult the <a href="https://user.eumetsat.int/resources/user-guides/sentinel-3-olci-ocean-colour-processing-baseline" target="_blank">OLCI level 2 product notices</a> for the relevant processing baseline.

```python
# reading masks: example land
land_mask = eumartools.flag_mask(flag_file, 'WQSF', ['LAND'])
land_mask = land_mask.astype(float)[::grid_reduce, ::grid_reduce]
land_mask[land_mask == 0.0] = np.nan

# reading masks: example cloud
cloud_mask = eumartools.flag_mask(flag_file, 'WQSF', ['CLOUD', 'CLOUD_AMBIGUOUS', 'CLOUD_MARGIN'])
cloud_mask = cloud_mask.astype(float)[::grid_reduce, ::grid_reduce]
cloud_mask[cloud_mask == 0.0] = np.nan

# reading masks: all flags NN recommended
NN_mask = eumartools.flag_mask(flag_file, 'WQSF', ['LAND', 'CLOUD', 'CLOUD_AMBIGUOUS', 'CLOUD_MARGIN', 
                                                    'INVALID', 'COSMETIC', 'SATURATED', 'SUSPECT',
                                                    'HISOLZEN', 'HIGHGLINT', 'SNOW_ICE', 'OCNN_FAIL'])
NN_mask = NN_mask.astype(float)[::grid_reduce, ::grid_reduce]
NN_mask[NN_mask == 0.0] = np.nan
```

<div class="alert alert-info" role="alert">

## Plotting and comparing constituents
[Back to top](#TOC_TOP_1_7)

</div>

Now that we've read in both the data we're interested in, and the flags to creat a mask, we can plot the data on a map...

```python
fig, axs = plt.subplots(1, 3, figsize=(20, 20), dpi=150, subplot_kw={"projection": ccrs.Mercator()})

for m, var, cmap, clabel in zip(axs.reshape(-1),
                         ["chl_nn", "tsm_nn", "adg_443_nn"],
                         ["viridis", "cividis", "plasma"], 
                         ["Chlorophyll concentration [mg.m$^{-3}$]", "Total suspended sediment [mg.m$^{-3}$]", "CDM absorption coefficient [m$^{-1}$]"]):
    
    p1 = m.pcolormesh(lon, lat, band_var[var][::grid_reduce, ::grid_reduce], transform=ccrs.PlateCarree(), 
                       cmap=cmap, vmin=-2, vmax=2, zorder=1)

    m.contourf(lon, lat, NN_mask, levels=[0,1], transform=ccrs.PlateCarree(), colors='0.75', zorder=2)
    m.contourf(lon, lat, land_mask, levels=[0,1], transform=ccrs.PlateCarree(), colors='0.5', zorder=3)
    m.contourf(lon, lat, cloud_mask, levels=[0,1], transform=ccrs.PlateCarree(), colors='0.25', zorder=4)

    # Embellish with gridlines
    g1 = m.gridlines(draw_labels = True, zorder=20, color='0.0', linestyle='--', linewidth=0.5)
    g1.xlocator = mticker.FixedLocator(np.arange(-180, 180, 5))
    g1.ylocator = mticker.FixedLocator(np.arange(-90, 90, 5))
    g1.top_labels = False
    g1.right_labels = False
    g1.xlabel_style = {'color': 'black'}
    g1.ylabel_style = {'color': 'black'}
    m.set(facecolor = "1.0")
    m.axis('off')

    cbar = fig.colorbar(p1, ax=m, location='bottom', pad=0.05)
    ticks = [-2,-1,0,1,2]
    cbar.set_ticks(ticks)
    cbar.set_ticklabels([10**tick for tick in ticks])
    cbar.set_label(clabel)

plt.tight_layout()
plt.savefig('1_7_OLCI_optical_constituents.png', bbox_inches='tight')
```
