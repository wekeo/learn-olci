### Outline

The colour of natural waters is influenced by the presence and concentration of a range of optically active constituents. By interacting with sunlight through absorption and scattering, these constituents alter the spectral characteristics of the light leaving the water and observed by satellite ocean colour instruments. This provides the basis for estimating their concentrations and distributions from remotely sensed ocean colour data.

Among the key in-water constituents that can be derived from ocean colour observations are chlorophyll-a (CHL), total suspended matter (TSM), and coloured dissolved matter (CDM). Chlorophyll-a provides an indication of phytoplankton abundance, whilst total suspended matter represents particles suspended within the water column, including sediments and other particulate material. Coloured dissolved matter, often associated with dissolved organic material, strongly absorbs light at shorter wavelengths and can have a significant influence on water colour.

The relative contributions of these constituents to the observed reflectance spectrum vary considerably between different water bodies and environmental conditions. In the open ocean, variations in water colour are often dominated by changes in phytoplankton and their associated constituents. In coastal and inland waters, however, suspended sediments and dissolved organic matter may also make substantial, and sometimes dominant, contributions to the observed signal. Separating these overlapping optical influences and accurately estimating individual constituent concentrations is therefore a significant challenge in ocean colour remote sensing.

In this notebook we'll explore the retrieval of chlorophyll-a, total suspended matter, and absorption due to coloured dissolved matter from ocean colour observations. We'll examine how these different constituents influence the observed water-leaving reflectance, how their concentrations are estimated from satellite data, and some of the challenges and limitations associated with interpreting these products across different water types.

## OLCI optical consitutents
[Back to top](#TOC_TOP_1_7)

</div>

Concentrations and optical properties of several in-water constituents are provided in the OLCI Level-2 products as outputs of the alternative atmospheric correction and complex water processing chain. You can find detailed information about these products in the relevant pages of the latest <a href="https://user.eumetsat.int/resources/user-guides/sentinel-3-ocean-colour-level-2-data-guide" target="_blank">Sentinel-3 ocean colour Level-2 data guide</a>.

The three products considered here are:

* CHL_NN
* TSM_NN
* ADG443_NN

These products are generated using a neural-network-based processing approach designed to retrieve water-leaving reflectances and associated geophysical properties, particularly in optically complex waters where several constituents may contribute independently to the observed ocean colour signal.

The chlorophyll-a concentration (CHL_NN) provides an estimate of the concentration of the photosynthetic pigment chlorophyll-a. The total suspended matter concentration (TSM_NN) provides an estimate of the concentration of particulate material suspended within the water column. The absorption coefficient ADG443_NN represents the combined absorption by coloured dissolved and detrital matter at a wavelength of 443 nm. Unlike CHL_NN and TSM_NN, which are expressed as concentrations, ADG443_NN is an inherent optical property and is expressed as an absorption coefficient.

The retrieval of these parameters is interconnected because chlorophyll-a, suspended matter, and coloured dissolved and detrital matter can all influence the spectral reflectance observed by an ocean colour instrument. Their relative contributions vary between water types, making it challenging to separate the individual constituents directly from the satellite-observed signal. The neural networks used in the complex water processing chain are trained to relate spectral reflectance to the optical properties and concentrations of these different constituents.

The complex water processing chain also includes an alternative atmospheric correction, which retrieves water-leaving reflectances that are used as inputs to the neural networks. These reflectances are not supplied as part of the standard Level-2 OLCI products, but the resulting geophysical products, including CHL_NN, TSM_NN, and ADG443_NN, are provided.

You can find additional information about Sentinel-3 OLCI level 2 products in the <a href="https://user.eumetsat.int/resources/user-guides/sentinel-3-ocean-colour-level-2-data-guide" target="_blank">Sentinel-3 ocean colour level 2 data guide</a> and the Sentinel-3 OLCI level 2 baseline collection 004 <a href="https://user.eumetsat.int/s3/ope-eup-strapi-media/S3_PN_OLCI_L2_M_004_00_Sentinel_3_Product_Notice_OLCI_Level_2_Water_v1_cfa434a623.pdf" target="_blank">product notice</a>. The complex water processor (C2RCC) is also available through the <a href="https://step.esa.int/main/toolboxes/snap/" target="_blank">Sentinel Applications Platform (SNAP)</a>.

## Reading OLCI optical consitutent data
[Back to top](#TOC_TOP_1_7)

</div>

To begin with, we'll create paths that point to the location of our data of interest.

```python
baseline_collection = '004'
SAFE_directory = glob.glob(os.path.join(os.getcwd(), 'products', f'*WFR*{baseline_collection}*'))[0]
```

Next we'll set a parameter to reduce the resolution of the data we are working with. For analyses you would not want to do this, but it will help us to visualise our data much faster in the context of this tutorial notebook.

```python
grid_reduce = 5
```

Then, we'll open our latitude and longitude grid, subsampling by `grid_reduce`.

```python
geo_file = os.path.join(SAFE_directory,'geo_coordinates.nc')
geo_fid = xr.open_dataset(geo_file)
lon = geo_fid.longitude.data[::grid_reduce, ::grid_reduce]
lat = geo_fid.latitude.data[::grid_reduce, ::grid_reduce]
geo_fid.close()
```

Lastly, we will open our three optical consistuent products, as provided by the complex water processor:
* the neural network chlorophyll product (CHL_NN; as discuss in the previous notebook)
* the neural network total suspended sediment product (TSM_NN)
* the neural network coloured detrital matter (CDM) absorption coefficient at 443 nm (ADG443_NN)

```python
band_var = {}
band_var["chl_nn"] = xr.open_mfdataset(glob.glob(os.path.join(SAFE_directory,'chl_nn.nc')))["CHL_NN"]
band_var["tsm_nn"] = xr.open_mfdataset(glob.glob(os.path.join(SAFE_directory,'tsm_nn.nc')))["TSM_NN"]
band_var["adg_443_nn"] = xr.open_mfdataset(glob.glob(os.path.join(SAFE_directory,'iop_nn.nc')))["ADG443_NN"]
```

<div class="alert alert-info" role="alert">

## Applying flags
[Back to top](#TOC_TOP_1_7)

</div>

As well as the data and the geocoordinates, to plot geophysical products derived from OLCI properly, we also need to extract and apply the relevant quality flags. Below you can see that the flags are extracted from the wqsf netcdf file. If we print out the flags and their associated bit values, you can get an idea about what they represent, and how to access each individual one.

```python
flag_file = os.path.join(SAFE_directory,'wqsf.nc')
```

We'll use a quick function from the `eumartools` library that allows us to read in the flag information simply by supplying the name. For our recommendations on which flags to apply you should also consult the <a href="https://user.eumetsat.int/resources/user-guides/sentinel-3-olci-ocean-colour-processing-baseline" target="_blank">OLCI level 2 product notices</a> for the relevant processing baseline.

```python
# reading masks: example land
land_mask = eumartools.flag_mask(flag_file, 'WQSF', ['LAND'])
land_mask = land_mask.astype(float)[::grid_reduce, ::grid_reduce]
land_mask[land_mask == 0.0] = np.nan

# reading masks: example cloud
cloud_mask = eumartools.flag_mask(flag_file, 'WQSF', ['CLOUD', 'CLOUD_AMBIGUOUS', 'CLOUD_MARGIN'])
cloud_mask = cloud_mask.astype(float)[::grid_reduce, ::grid_reduce]
cloud_mask[cloud_mask == 0.0] = np.nan
```

We'll create a combined mask by reading in all the flags that are relevant for both the products derived from the neural network processor. These are the recommended flags provided by EUMETSAT.

```python
# reading masks: all flags NN recommended
NN_mask = eumartools.flag_mask(flag_file, 'WQSF', ['LAND', 'CLOUD', 'CLOUD_AMBIGUOUS', 'CLOUD_MARGIN', 
                                                    'INVALID', 'COSMETIC', 'SATURATED', 'SUSPECT',
                                                    'HISOLZEN', 'HIGHGLINT', 'SNOW_ICE', 'OCNN_FAIL'])
NN_mask = NN_mask.astype(float)[::grid_reduce, ::grid_reduce]
NN_mask[NN_mask == 0.0] = np.nan
```

<div class="alert alert-info" role="alert">

## Plotting and comparing constituents
[Back to top](#TOC_TOP_1_7)

</div>

Now that we've read in both the data we're interested in, and the flags to creat a mask, we can plot the data on a map...

```python
fig, axs = plt.subplots(1, 3, figsize=(20, 20), dpi=150, subplot_kw={"projection": ccrs.Mercator()})

for m, var, cmap, clabel in zip(axs.reshape(-1),
                         ["chl_nn", "tsm_nn", "adg_443_nn"],
                         ["viridis", "cividis", "plasma"], 
                         ["Chlorophyll concentration [mg.m$^{-3}$]", "Total suspended sediment [mg.m$^{-3}$]", "CDM absorption coefficient [m$^{-1}$]"]):
    
    p1 = m.pcolormesh(lon, lat, band_var[var][::grid_reduce, ::grid_reduce], transform=ccrs.PlateCarree(), 
                       cmap=cmap, vmin=-2, vmax=2, zorder=1)

    m.contourf(lon, lat, NN_mask, levels=[0,1], transform=ccrs.PlateCarree(), colors='0.75', zorder=2)
    m.contourf(lon, lat, land_mask, levels=[0,1], transform=ccrs.PlateCarree(), colors='0.5', zorder=3)
    m.contourf(lon, lat, cloud_mask, levels=[0,1], transform=ccrs.PlateCarree(), colors='0.25', zorder=4)

    # Embellish with gridlines
    g1 = m.gridlines(draw_labels = True, zorder=20, color='0.0', linestyle='--', linewidth=0.5)
    g1.xlocator = mticker.FixedLocator(np.arange(-180, 180, 5))
    g1.ylocator = mticker.FixedLocator(np.arange(-90, 90, 5))
    g1.top_labels = False
    g1.right_labels = False
    g1.xlabel_style = {'color': 'black'}
    g1.ylabel_style = {'color': 'black'}
    m.set(facecolor = "1.0")
    m.axis('off')

    cbar = fig.colorbar(p1, ax=m, location='bottom', pad=0.05)
    ticks = [-2,-1,0,1,2]
    cbar.set_ticks(ticks)
    cbar.set_ticklabels([10**tick for tick in ticks])
    cbar.set_label(clabel)

plt.tight_layout()
plt.savefig('1_7_OLCI_optical_constituents.png', bbox_inches='tight')
```
