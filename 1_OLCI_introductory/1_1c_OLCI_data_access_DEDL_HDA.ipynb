### DestinE Data Lake

The DestinE Data Lake (DEDL) provides discovery, access, and big data processing services, supporting the DestinE Service Platform (DESP). Access to the data stored within the DEDL is made possible via the Destination Earth Data Lake Harmonised Data Access (HDA) API, which provides seamless access to both “DestinE Data” and “Federated Data” as defined in the <a href="https://destine-data-lake-docs.data.destination-earth.eu/en/latest/dedl-discovery-and-data-access/DestinE-Data-Portfolio/DestinE-Data-Portfolio.html" target="_blank">DestinE Data Portfolio</a>. The data catalogue includes Copernicus Sentinel-3 and Senitnel-6 marine products. The API is based on the Spatio Temporal Asset Catalog specification (STAC) allowing users to discover datasets and services. Datasets contain data which can be accessed or downloaded for further processing. HDA exposes a set of RESTful APIs for programmatic access to data, along with a STAC 1.0.0-compliant interface.

### Accessing the DestinE Data Lake

To download Copernicus marine products from the <a href="https://destination-earth.eu/destination-earth/destines-components/destination-earth-data-lake/" target="_blank">DestinE Data Lake</a> (DEDL), we will use the DestinE <a href="https://platform.destine.eu/services/service/hda/" target="_blank">Harmonised Data Access (HDA)</a> adaptor. If you are currently working on the DEDL JupyterHub, this client should already be installed. Otherwise, you can use the environment file included in this repository (environment.yml) to build your own python environment (as detailed in the README) and install it.

In order to allow us to download data using the DEDL HDA adaptor, we need to provide our credentials. To do this, we need to create a file called `.credentials` in our folder called `.destine` in our home directory. For most computer systems the home directory can be found at the path \user\username, /users/username, or /home/username depending on your operating system. In this file we need to add the following information exactly as follows;

```
username,password
```
<br>

You must replace `<username>` and `<password>` with the information from your DestinE account (if you don't have one yet, please register at <a href=https://platform.destine.eu/login/ target="_blank">https://www.wekeo.eu/</a>.

</div>

Lets create our credentials file, if it doesn't already exist, and read our credentials from it.

````python
# load credentials
destine_credentials_file = Path(Path.home() / '.destine' / 'credentials')

if os.path.exists(destine_credentials_file):
    username, password = Path(destine_credentials_file).read_text().split(',')
else:
    # creating authentication file
    username = input('Enter your DestinE DESP username: ')
    password = getpass.getpass('Enter your DestinE DESP password: ')
    try:
        os.makedirs(os.path.dirname(destine_credentials_file), exist_ok=True)
        with open(destine_credentials_file, "w") as f:
            f.write(f'{username},{password}')
    except:
        pass
        
access_token = destinelab.AuthHandler(username, password).get_token()
auth_headers = {"Authorization": f"Bearer {access_token}"}
````

The DEDL HDA allows you to search for data in time and space, but not by product name. However, we can work around this by providing the exact time stamps for our data. Below, we have a query for the first item in our product list, the OLCI level 1B product. You can see that the start and end times match the filename times, with a single second removed from the end time so that we don't get the next product.

````python
query_EFR = {
    "collections": ["EO.EUM.DAT.SENTINEL-3.OL_1_EFR___"],
    "datetime": "2025-04-03T10:16:16.000Z/2025-04-03T10:19:14.000Z"}
````

Lets now make our search, to check for matching products...

````python
response = requests.post("https://hda.data.destination-earth.eu/stac/search", headers=auth_headers, json=query_EFR)
````

Lets see what we found. If all went well, it should be a single OLCI L1B product. This will be stored under the "0" menu item in the cell below...

````python
JSON(response.json()["features"])
````

Lets now download this product.

````python
product = response.json()["features"][0]
response = requests.get(product["assets"]["downloadLink"]["href"], headers=auth_headers, stream=True)
with open(product["id"], 'wb') as f:
    for chunk in response.iter_content(chunk_size=1024):
        f.write(chunk)
````

Now, we can unzip the file into our download directory.

````python
with zipfile.ZipFile(product["id"], 'r') as zip_ref:
    zip_ref.extractall(download_dir)
    print(f'Unzipping of product {product["id"]} finished.')
os.remove(product["id"])
````

Now lets download our other products. We will need to set up query files for them, and download in a loop.

````python
query_WRR = {
    "collections": ["EO.EUM.DAT.SENTINEL-3.OL_2_WRR___"],
    "datetime": "2025-04-03T10:08:28.000Z/2025-04-03T10:52:33.000Z"}

query_WFR_1 = {
    "collections": ["EO.EUM.DAT.SENTINEL-3.OL_2_WFR___"],
    "datetime": "2025-04-03T10:16:16.000Z/2025-04-03T10:19:14.000Z"}

query_WFR_2 = {
    "collections": ["EO.EUM.DAT.SENTINEL-3.OL_2_WFR___"],
    "datetime": "2026-03-05T10:05:18.000Z/2026-03-05T10:08:16.000Z"}
    
queries = [query_WRR, query_WFR_1, query_WFR_2]
````

And again we can search for, and download the results...

````python
for query in queries:
    response = requests.post("https://hda.data.destination-earth.eu/stac/search", headers=auth_headers, json=query)
    product = response.json()["features"][0]
    response = requests.get(product["assets"]["downloadLink"]["href"], headers=auth_headers, stream=True)
    with open(product["id"], 'wb') as f:
        for chunk in response.iter_content(chunk_size=1024):
            f.write(chunk)
    with zipfile.ZipFile(product["id"], 'r') as zip_ref:
        zip_ref.extractall(download_dir)
        print(f'Unzipping of product {product["id"]} finished.')
    os.remove(product["id"])
````
